```python
import ee

ee.Initialize(project='gee-mb-06')
```

```python
lon_min, lat_min, lon_max, lat_max = -55.0, -5.0, -54.9, -4.9
aoi = ee.Geometry.Rectangle([lon_min, lat_min, lon_max, lat_max])

collection = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
              .filterBounds(aoi)
              .filterDate('2024-01-01', '2024-12-31')
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 60)))

image = collection.median().clip(aoi)
```

```python
worldcover = ee.ImageCollection("ESA/WorldCover/v200").first()
dem = ee.Image("USGS/SRTMGL1_003")
gedi = ee.Image('LARSE/GEDI/GEDI04_B_002')
```

```python
import geemap

Map = geemap.Map()
Map.centerObject(aoi, 10)


vis_params_falsecolor = {'bands': ['B8', 'B4', 'B3'], 'min': 0, 'max': 3000}
Map.addLayer(image, vis_params_falsecolor, 'False Color (NIR)')

Map.addLayer(aoi, {}, 'AOI', False)
Map
```

```python
import geemap

Map = geemap.Map()
Map.centerObject(aoi, 10)


vis_params = {'bands': ['B4', 'B3', 'B2'], 'min': 0, 'max': 2000, 'gamma': 1.4}
Map.addLayer(image, vis_params, 'Sentinel-2 Composite')

Map.addLayer(aoi, {}, 'AOI', False)
Map
```
